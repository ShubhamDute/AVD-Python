# multiplying in a list vs an array

```python
# if you directly multiply by 4 to a list, list will repeat itself
a = [5, 6, 7, 8, 9, 10]
print(4 * a)
```

```python
1.5 * a
```

```python
a = [5, 6, 7, 8, 9, 10]
print(a)

# Initialize blank results list

r = []

# Apply for loop on a
for i in a:
    print(f"{i} x 4 = {4 * i}")
    r.append(4 * i)
```

```python
print(r)
```

```python
a1 = np.array(a)
print(a1, type(a1))
b1 = 4 * a1
print(b1) 
```

```python
c1 = 5 * a1 + 3
print(c1)
```

```python
a1 ** 2
```

```python
a1 ** 3
```

```python
a1 ** (1/2)
```

# Sum of elements inside a list vs array

```python
# list 
b = [4 , 5, 6, 7, 8, 9, 10, 11, 12]
print(b)


# initialize sum with 0

s = 0

for i in b:
    s  = s + i


print(f"Sum of elements inside list  {b} is {s}")    
```

```python
# Array 
b1 = np.array(b)
b1.sum()
```

```python
b1.prod()
```

```python
b1.mean()
```

```python
b1.min()
```

```python
b1.max()
```

# Simulate normal distribution

```python
height = np.random.normal(loc = 5.5, scale=0.5/3, size=100000)
print(height)
print(height.shape)
```

```python
import seaborn as sns
sns.histplot(height, kde=True)
```
