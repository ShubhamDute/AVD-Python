# with exception handling

```python
def division2(num : int | float, den : int | float )-> float:
    try:
        return num / den
    except Exception as e:
        print(f"Exception Occured : {e}")
```

```python
d1 = division2(num= 21, den=4)
print(d1)
print("Done")
```

```python
d2 = division2(24, 0)
print(d2)
print("Done")
```

```python
d3 = division2("three", 5)
print(d3)
print("Done")
```

# custom exception handling
1. ZeroDivision Error
2. Type Error

```python
def division3(num : int | float, den : int | float)-> float:
    try:
        return num / den
    except ZeroDivisionError:
        print("Denometer cannot be 0 in division")
    except TypeError:
        print("Numerator and denometer should be a number")        
```

```python
r1 = division3(100, 3)
print(r1)
print("Done")
```

```python
r2 = division3(54, 0)
print(r2)
print("Done")
```

```python
r4 = division3("Five" , 7)
print(r4)
print("Done")
```

```python
r5 = division3("a", "b")
print(r5)
print("Done")
```

# Raise custom error
1. i want to take age as input from user
2. age should be between 0 to 100

```python
def age_validator():
    age = int(input("Please enter age in years : "))
    if age <= 0 or age > 100:
       raise ValueError("Age must be between 1 to 100")
    return age
```

```python
# With try except block
try:
    age2 = age_validator()
    print(f"valid age entered :{age2}")
except Exception as e:
    print(f"Exception Occured : {e}")    
```

# Different exception handling blocks


```python
import math

math.sqrt(25)
```

```python
math.sqrt(12)
```

```python
math.sqrt(-1)
```

```python
def square_root(num : int | float):
    try:
        s = math.sqrt(num)
    except Exception as e:
        print(f"Exception Occured : {e}")

    else:
        print(f"Square root of : {num} is {s}")
        return s
    finally:
        print("Finally block will always be excuted")        
```

```python
s1 = square_root(23)
```

```python
s2 = square_root(-23)
```

```python
s3 = square_root("abc")
```
