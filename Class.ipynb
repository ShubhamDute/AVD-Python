# Create a class person

1. attributes - properties of the class
2. methods  - perform certain operations based on attribute  

```python
class Person:
    # constructor functions _init_ to define attributes of class
    # Self is any object of class person
    def __init__(self, name: str, age: int, gender: str, occupation: str):
        self.name = name
        self.age = age
        self.gender = gender
        self.occupation = occupation

    # this function is the string representation of the class
    def __str__(self):
        return f"Person(name={self.name}, age={self.age}, gender={self.gender}, occupation={self.occupation})"

    # write a method for a class person
    def introduce(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
        print(f"I am {self.gender}")
        print(f"I work as {self.occupation}")        
```

```python
# p1 is object of class person
p1 = Person(name="shubham", age=23, gender="male", occupation="data analyst")
print(type(p1))
```

# data classes

```python
from dataclasses import dataclass
from typing import Literal

@dataclass
class Person2:
    name: str
    age: int
    gender: Literal["male", "female"]
    occupation: str

    def intro(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
        print(f"I am {self.gender}")
        print(f"I work as {self.occupation}")
```

```python
p3 = Person2(name="shubham", age=23, gender="male", occupation="data analyst")
print(p3)
```

```python
@dataclass
class Account:
    ac_no: int
    name: str
    balance: float

    def check_balance(self):
        print(f"Account number {self.ac_no}, Name : {self.name}, Balance : {self.balance}")

    def deposit(self, amt: float):
        if amt<=0:
            raise ValueError("The amount deposited cannot be negative or zero")
        self.balance = self.balance +amt

    def withdraw(self, amt: float):
        if amt <=0:
            raise ValueError("The ammount withdrawn cannot be negative or zero")
        if amt > self.balance:
            raise ValueError("Insufficient banlance")
        self.balance = self.balance - amt       

    def transfer(self, amt: float, other_ac: 'Account'):
        self.withdraw(amt)
        other_ac.deposit(amt)    
```

```python
ac1 = Account(ac_no= 1342, name= "shubham", balance=1000000)
print(ac1)
```

```python
ac1.check_balance()
```

```python
ac1.withdraw(100001)
```

```python
ac1.balance
```
